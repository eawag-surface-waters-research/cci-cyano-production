```python
import os
import netCDF4
import numpy as np
from datetime import datetime
import matplotlib.pyplot as plt
```

```python
file = "../data/v2.1/extract/chla_mean/327.nc"
```

Timeseries at one pixel

```python
with netCDF4.Dataset(file) as nc:
    x = nc.variables["chla_mean"][:, 30, 50]
    t = nc.variables["time"]
    time = [datetime.fromtimestamp(int(ts)) for ts in np.array(t)]
    plt.plot(time, x)
    plt.show()
```

At specific time point

```python
with netCDF4.Dataset(file) as nc:
    x = nc.variables["chla_mean"][0, :, :]
    plt.imshow(x)
    plt.colorbar()
    plt.show()
```

Number of values in dataset

```python
with netCDF4.Dataset(file) as nc:
    x = nc.variables["summary"][:, :]

    print(len(x[x!=0]))
    plt.imshow(x)
    plt.colorbar()
    plt.show()
```
